### Block 7: Functional tools
This block demonstrates `functools.partial` for fixing arguments of a function, plus `map` and `filter` as alternatives to list comprehensions.

```python
from functools import partial

def exp(base, power):
    """base raised to the power'th power"""
    return base ** power

# fixing the first argument gives us a new one-argument function
two_to_the = partial(exp, 2)
print(two_to_the(3))   # 2 ** 3

# we can also fix a keyword argument
square_of = partial(exp, power=2)
print(square_of(5))    # 5 ** 2

def double(x):
    return 2 * x

xs = [1, 2, 3, 4]
twice_xs = [double(x) for x in xs]
print(twice_xs)

twice_xs_map = list(map(double, xs))
print(twice_xs_map)

list_doubler = partial(map, double)
print(list(list_doubler(xs)))

def multiply(x, y):
    return x * y

products = list(map(multiply, [1, 2, 3], [4, 5, 6]))
print(products)

def is_even(x):
    return x % 2 == 0

x_evens = [x for x in xs if is_even(x)]
print(x_evens)
x_evens_filter = list(filter(is_even, xs))
print(x_evens_filter)
```

### Block 9: args and kwargs
This block demonstrates variadic functions using `*args`/`**kwargs`, unpacking a list and dict into a call, and writing a `doubler` decorator-like wrapper that only works correctly once it also accepts `*args, **kwargs`.

```python
def magic(*args, **kwargs):
    print("unnamed args:", args)
    print("keyword args:", kwargs)

magic(1, 2, key="word", key2="word2")

def other_way_magic(x, y, z):
    return x + y + z

x_y_list = [1, 2]
z_dict = {"z": 3}
print(other_way_magic(*x_y_list, **z_dict))

def doubler(f):
    def g(x):
        return 2 * f(x)
    return g

def f1(x):
    return x + 1

g = doubler(f1)
print(g(3))   # 2 * (3 + 1) = 8
print(g(-1))  # 2 * (-1 + 1) = 0

def f2(x, y):
    return x + y

g2 = doubler(f2)
try:
    g2(1, 2)
except TypeError:
    print("as defined, g only takes one argument")

def magic_doubler(f):
    def g(*args, **kwargs):
        return 2 * f(*args, **kwargs)
    return g

g3 = magic_doubler(f2)
print(g3(1, 2))
```

### Block 11: A first class (basic OOP)
This block defines a simple `CountingClicker` class with `__init__`, `__repr__`, and methods, then subclasses it to override behavior — a minimal example of encapsulation and inheritance.

```python
class CountingClicker:
    """A simple counter class to demonstrate basic OOP in Python."""

    def __init__(self, count: int = 0) -> None:
        self.count = count

    def __repr__(self) -> str:
        return f"CountingClicker(count={self.count})"

    def click(self, num_times: int = 1) -> None:
        """Click the counter some number of times."""
        self.count += num_times

    def read(self) -> int:
        return self.count

    def reset(self) -> None:
        self.count = 0


clicker = CountingClicker()
print(clicker)
clicker.click()
clicker.click(5)
print(clicker.read())
clicker.reset()
print(clicker.read())


class NoResetClicker(CountingClicker):
    """A subclass that overrides reset() to do nothing."""
    def reset(self) -> None:
        pass

clicker2 = NoResetClicker()
clicker2.click(3)
clicker2.reset()
print(clicker2.read())
```

### Block 15: Vectors from scratch
This block implements vectors as plain lists of numbers and builds up `add`, `subtract`, `vector_sum`, `scalar_multiply`, `vector_mean`, `dot`, `sum_of_squares`, `magnitude`, and `distance` from first principles (no NumPy).

```python
from typing import List
import math

Vector = List[float]

def add(v: Vector, w: Vector) -> Vector:
    """Adds corresponding elements"""
    assert len(v) == len(w), "vectors must be the same length"
    return [v_i + w_i for v_i, w_i in zip(v, w)]

def subtract(v: Vector, w: Vector) -> Vector:
    """Subtracts corresponding elements"""
    assert len(v) == len(w), "vectors must be the same length"
    return [v_i - w_i for v_i, w_i in zip(v, w)]

def vector_sum(vectors: List[Vector]) -> Vector:
    """Sums all corresponding elements"""
    assert vectors, "no vectors provided!"
    num_elements = len(vectors[0])
    assert all(len(v) == num_elements for v in vectors), "different sizes!"
    return [sum(vector[i] for vector in vectors) for i in range(num_elements)]

def scalar_multiply(c: float, v: Vector) -> Vector:
    """Multiplies every element by c"""
    return [c * v_i for v_i in v]

def vector_mean(vectors: List[Vector]) -> Vector:
    """Computes the element-wise average"""
    n = len(vectors)
    return scalar_multiply(1 / n, vector_sum(vectors))

def dot(v: Vector, w: Vector) -> float:
    """Computes v_1 * w_1 + ... + v_n * w_n"""
    assert len(v) == len(w), "vectors must be the same length"
    return sum(v_i * w_i for v_i, w_i in zip(v, w))

def sum_of_squares(v: Vector) -> float:
    """Returns v_1 * v_1 + ... + v_n * v_n"""
    return dot(v, v)

def magnitude(v: Vector) -> float:
    """Returns the magnitude (length) of v"""
    return math.sqrt(sum_of_squares(v))

def distance(v: Vector, w: Vector) -> float:
    """Computes the distance between v and w"""
    return magnitude(subtract(v, w))


v = [1, 2, 3]
w = [4, 5, 6]

print(add(v, w))
print(subtract(v, w))
print(vector_sum([v, w, [1, 1, 1]]))
print(scalar_multiply(2, v))
print(vector_mean([v, w]))
print(dot(v, w))
print(sum_of_squares(v))
print(magnitude(v))
print(distance(v, w))
```

### Block 16: Matrices from scratch
This block represents a matrix as a list of lists (rows), and writes `shape`, `get_row`, `get_column`, and `make_matrix` (used here to build an identity matrix and a friendship adjacency matrix).

```python
from typing import List, Tuple, Callable

Matrix = List[List[float]]

def shape(A: Matrix) -> Tuple[int, int]:
    """Returns (number of rows, number of columns) of A"""
    num_rows = len(A)
    num_cols = len(A[0]) if A else 0
    return num_rows, num_cols

def get_row(A: Matrix, i: int) -> List[float]:
    """Returns the i-th row of A (as a list)"""
    return A[i]

def get_column(A: Matrix, j: int) -> List[float]:
    """Returns the j-th column of A (as a list)"""
    return [A_i[j] for A_i in A]

def make_matrix(num_rows: int,
                 num_cols: int,
                 entry_fn: Callable[[int, int], float]) -> Matrix:
    """
    Returns a num_rows x num_cols matrix
    whose (i, j)-th entry is entry_fn(i, j)
    """
    return [[entry_fn(i, j) for j in range(num_cols)]
            for i in range(num_rows)]

def identity_matrix(n: int) -> Matrix:
    """Returns the n x n identity matrix"""
    return make_matrix(n, n, lambda i, j: 1 if i == j else 0)


A = [[1, 2, 3],
     [4, 5, 6]]

print(shape(A))
print(get_row(A, 1))
print(get_column(A, 2))
print(identity_matrix(5))

# a matrix can represent a dataset: each row is an observation,
# each column is a feature
friendships = [(0, 1), (1, 2), (2, 3)]
n = 4
friend_matrix = make_matrix(n, n, lambda i, j: 1 if (i, j) in friendships or (j, i) in friendships else 0)
for row in friend_matrix:
    print(row)
```

### Block 17: Measures of central tendency
This block computes `mean`, `median` (handling both even and odd length lists), `quantile`, and `mode` on a sample dataset of "number of friends" values.

```python
from typing import List
from collections import Counter

num_friends = [100, 49, 41, 40, 25, 21, 21, 19, 19, 18, 18, 16, 15, 15, 15,
               15, 14, 14, 13, 13, 13, 13, 12, 12, 11, 10, 10, 10, 10, 9,
               9, 9, 9, 9, 8, 8, 8, 6, 6, 5, 5, 4, 4, 4, 3, 3, 3, 3, 2, 2, 1]

def mean(xs: List[float]) -> float:
    return sum(xs) / len(xs)

def _median_odd(xs: List[float]) -> float:
    return sorted(xs)[len(xs) // 2]

def _median_even(xs: List[float]) -> float:
    sorted_xs = sorted(xs)
    mid = len(xs) // 2
    return (sorted_xs[mid - 1] + sorted_xs[mid]) / 2

def median(v: List[float]) -> float:
    """Finds the 'middle-most' value of v"""
    return _median_even(v) if len(v) % 2 == 0 else _median_odd(v)

def quantile(xs: List[float], p: float) -> float:
    """Returns the pth-percentile value in x"""
    p_index = int(p * len(xs))
    return sorted(xs)[p_index]

def mode(x: List[float]) -> List[float]:
    """Returns a list, since there might be more than one mode"""
    counts = Counter(x)
    max_count = max(counts.values())
    return [x_i for x_i, count in counts.items() if count == max_count]


print(mean(num_friends))
print(median(num_friends))
print(median([1, 10, 2, 9, 5]))
print(median([1, 9, 2, 10]))
print(quantile(num_friends, 0.10))
print(quantile(num_friends, 0.25))
print(quantile(num_friends, 0.75))
print(quantile(num_friends, 0.90))
print(mode(num_friends))
```

### Block 18: Dispersion
This block computes spread-of-data statistics: `data_range`, `variance` (via `de_mean` and `sum_of_squares`), `standard_deviation`, and `interquartile_range`.

```python
from typing import List
import math

num_friends = [100, 49, 41, 40, 25, 21, 21, 19, 19, 18, 18, 16, 15, 15, 15,
               15, 14, 14, 13, 13, 13, 13, 12, 12, 11, 10, 10, 10, 10, 9,
               9, 9, 9, 9, 8, 8, 8, 6, 6, 5, 5, 4, 4, 4, 3, 3, 3, 3, 2, 2, 1]

def mean(xs: List[float]) -> float:
    return sum(xs) / len(xs)

def data_range(xs: List[float]) -> float:
    return max(xs) - min(xs)

def de_mean(xs: List[float]) -> List[float]:
    """Translate xs by subtracting its mean (so the result has mean 0)"""
    x_bar = mean(xs)
    return [x - x_bar for x in xs]

def dot(v: List[float], w: List[float]) -> float:
    return sum(v_i * w_i for v_i, w_i in zip(v, w))

def sum_of_squares(v: List[float]) -> float:
    return dot(v, v)

def variance(xs: List[float]) -> float:
    """Almost the average squared deviation from the mean"""
    assert len(xs) >= 2, "variance requires at least two elements"
    n = len(xs)
    deviations = de_mean(xs)
    return sum_of_squares(deviations) / (n - 1)

def standard_deviation(xs: List[float]) -> float:
    """The standard deviation is the square root of the variance"""
    return math.sqrt(variance(xs))

def interquartile_range(xs: List[float]) -> float:
    """Returns the difference between the 75%-ile and the 25%-ile"""
    sorted_xs = sorted(xs)
    p75 = sorted_xs[int(0.75 * len(xs))]
    p25 = sorted_xs[int(0.25 * len(xs))]
    return p75 - p25


print(data_range(num_friends))
print(variance(num_friends))
print(standard_deviation(num_friends))
print(interquartile_range(num_friends))
```

### Block 19: Correlation
This block computes `covariance` and `correlation` between two variables, showing how strongly they move together on a scale of -1 to 1.

```python
from typing import List

num_friends = [100, 49, 41, 40, 25, 21, 21, 19, 19, 18, 18, 16, 15, 15, 15,
               15, 14, 14, 13, 13, 13, 13, 12, 12, 11, 10, 10, 10, 10, 9,
               9, 9, 9, 9, 8, 8, 8, 6, 6, 5, 5, 4, 4, 4, 3, 3, 3, 3, 2, 2, 1]

# pretend this is "minutes spent on the site per day" for the same 50 users
daily_minutes = [1 + 0.9 * x + ((-1) ** i) * (x % 5) for i, x in enumerate(num_friends)]

def mean(xs: List[float]) -> float:
    return sum(xs) / len(xs)

def de_mean(xs: List[float]) -> List[float]:
    x_bar = mean(xs)
    return [x - x_bar for x in xs]

def dot(v: List[float], w: List[float]) -> float:
    return sum(v_i * w_i for v_i, w_i in zip(v, w))

def sum_of_squares(v: List[float]) -> float:
    return dot(v, v)

def variance(xs: List[float]) -> float:
    n = len(xs)
    deviations = de_mean(xs)
    return sum_of_squares(deviations) / (n - 1)

def standard_deviation(xs: List[float]) -> float:
    return variance(xs) ** 0.5

def covariance(xs: List[float], ys: List[float]) -> float:
    """Measures how much xs and ys vary in tandem about their means"""
    assert len(xs) == len(ys), "xs and ys must have the same number of elements"
    return dot(de_mean(xs), de_mean(ys)) / (len(xs) - 1)

def correlation(xs: List[float], ys: List[float]) -> float:
    """Measures how much xs and ys vary together, on a scale of -1 to 1"""
    stdev_x = standard_deviation(xs)
    stdev_y = standard_deviation(ys)
    if stdev_x > 0 and stdev_y > 0:
        return covariance(xs, ys) / (stdev_x * stdev_y)
    else:
        return 0  # if no variation, correlation is zero


print(covariance(num_friends, daily_minutes))
print(correlation(num_friends, daily_minutes))
```

### Block 20: Conditional probability (simulation)
This block simulates the classic "two children" probability puzzle with an `enum.Enum` for `Kid` and Monte Carlo sampling, estimating P(both girls | older is a girl) vs. P(both girls | at least one is a girl).

```python
import enum
import random

class Kid(enum.Enum):
    BOY = 0
    GIRL = 1

def random_kid() -> Kid:
    return random.choice([Kid.BOY, Kid.GIRL])

random.seed(0)

both_girls = 0
older_girl = 0
either_girl = 0

for _ in range(10000):
    younger = random_kid()
    older = random_kid()
    if older == Kid.GIRL:
        older_girl += 1
    if older == Kid.GIRL and younger == Kid.GIRL:
        both_girls += 1
    if older == Kid.GIRL or younger == Kid.GIRL:
        either_girl += 1

print("P(both | older):", both_girls / older_girl)     # should be ~1/2
print("P(both | either):", both_girls / either_girl)   # should be ~1/3
```

### Block 21: The uniform distribution
This block implements the PDF and CDF of the uniform distribution on [0, 1) from scratch.

```python
def uniform_pdf(x: float) -> float:
    """The uniform distribution puts equal weight on every value in [0, 1)"""
    return 1 if 0 <= x < 1 else 0

def uniform_cdf(x: float) -> float:
    """Returns the probability that a uniform random variable is <= x"""
    if x < 0:
        return 0
    elif x < 1:
        return x
    else:
        return 1

print(uniform_pdf(0.5), uniform_pdf(1.5))
print(uniform_cdf(-1), uniform_cdf(0.3), uniform_cdf(0.75), uniform_cdf(2))
```

### Block 22: The normal distribution
This block implements the normal distribution's PDF and CDF using `math.erf`, plus an `inverse_normal_cdf` found via binary search — useful for turning a probability back into a z-score.

```python
import math

SQRT_TWO_PI = math.sqrt(2 * math.pi)

def normal_pdf(x: float, mu: float = 0, sigma: float = 1) -> float:
    return (math.exp(-(x - mu) ** 2 / (2 * sigma ** 2)) / (SQRT_TWO_PI * sigma))

def normal_cdf(x: float, mu: float = 0, sigma: float = 1) -> float:
    """The probability that a normal random variable is <= x"""
    return (1 + math.erf((x - mu) / math.sqrt(2) / sigma)) / 2

def inverse_normal_cdf(p: float,
                        mu: float = 0,
                        sigma: float = 1,
                        tolerance: float = 0.00001) -> float:
    """Find approximate inverse using binary search"""
    if mu != 0 or sigma != 1:
        return mu + sigma * inverse_normal_cdf(p, tolerance=tolerance)

    low_z = -10.0
    hi_z = 10.0
    while hi_z - low_z > tolerance:
        mid_z = (low_z + hi_z) / 2
        mid_p = normal_cdf(mid_z)
        if mid_p < p:
            low_z = mid_z
        else:
            hi_z = mid_z
    return mid_z


print(normal_pdf(0))          # peak of the standard normal
print(normal_cdf(0))          # should be 0.5
print(normal_cdf(1.96))       # ~0.975
print(inverse_normal_cdf(0.975))  # ~1.96, sanity check against normal_cdf
```

### Block 23: The Central Limit Theorem
This block simulates a Binomial(n, p) random variable many times, plots its histogram, and overlays the normal distribution that approximates it as n grows — a visual demonstration of the Central Limit Theorem.

```python
import random
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from collections import Counter

def bernoulli_trial(p: float) -> int:
    """Returns 1 with probability p, 0 with probability 1-p"""
    return 1 if random.random() < p else 0

def binomial(n: int, p: float) -> int:
    """Returns the sum of n bernoulli(p) trials"""
    return sum(bernoulli_trial(p) for _ in range(n))

def make_hist(p: float, n: int, num_points: int) -> None:
    """Picks points from a Binomial(n, p) and plots their histogram"""
    data = [binomial(n, p) for _ in range(num_points)]

    # use a bar chart to show the actual binomial samples
    histogram = Counter(data)
    plt.bar([x - 0.4 for x in histogram.keys()],
            [v / num_points for v in histogram.values()],
            0.8,
            color="0.75")

    mu = p * n
    sigma = (n * p * (1 - p)) ** 0.5

    # use a line chart to show the normal approximation
    import math
    SQRT_TWO_PI = math.sqrt(2 * math.pi)
    def normal_pdf(x, mu=0, sigma=1):
        return math.exp(-(x - mu) ** 2 / (2 * sigma ** 2)) / (SQRT_TWO_PI * sigma)

    xs = range(min(data), max(data) + 1)
    ys = [normal_pdf(i, mu, sigma) for i in xs]
    plt.plot(xs, ys)
    plt.title("Binomial Distribution vs. Normal Approximation")
    plt.savefig("clt_demo.png", dpi=100, bbox_inches="tight")
    plt.close()

random.seed(0)
make_hist(0.75, 100, 10000)
print("saved chart")
```
